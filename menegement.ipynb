```python
%%writefile student-management/storage.py

import json

FILE_NAME = "student-management/data/mahasiswa.json"


def baca_data():
    try:
        with open(FILE_NAME, "r", encoding="utf-8") as file:
            return json.load(file)

    except FileNotFoundError:
        return []

    except json.JSONDecodeError:
        return []


def simpan_data(data):
    with open(FILE_NAME, "w", encoding="utf-8") as file:
        json.dump(data, file, indent=4, ensure_ascii=False)
```

```python
%%writefile student-management/mahasiswa.py

from storage import baca_data, simpan_data


def tambah_mahasiswa():
    data = baca_data()

    nama = input("Nama  : ")
    nim = input("NIM   : ")
    prodi = input("Prodi : ")

    while True:
        try:
            ipk = float(input("IPK   : "))

            if 0 <= ipk <= 4:
                break

            print("IPK harus antara 0 sampai 4.")

        except ValueError:
            print("IPK harus berupa angka.")

    mahasiswa_baru = {
        "nim": nim,
        "nama": nama,
        "prodi": prodi,
        "ipk": ipk
    }

    data.append(mahasiswa_baru)

    simpan_data(data)

    print("Data berhasil ditambahkan!")


def lihat_mahasiswa():
    data = baca_data()

    if not data:
        print("Belum ada data mahasiswa.")
        return

    print("\n===== DATA MAHASISWA =====")

    for mahasiswa in data:
        print("----------------------")
        print("NIM   :", mahasiswa["nim"])
        print("Nama  :", mahasiswa["nama"])
        print("Prodi :", mahasiswa["prodi"])
        print("IPK   :", mahasiswa["ipk"])


def cari_mahasiswa():
    data = baca_data()

    cari = input("Masukkan NIM: ")

    for mahasiswa in data:
        if mahasiswa["nim"] == cari:
            print("\nData ditemukan!")
            print("NIM   :", mahasiswa["nim"])
            print("Nama  :", mahasiswa["nama"])
            print("Prodi :", mahasiswa["prodi"])
            print("IPK   :", mahasiswa["ipk"])
            return

    print("Mahasiswa tidak ditemukan.")
```

```python
%%writefile student-management/analisis.py

from storage import baca_data


def rata_rata_ipk():
    data = baca_data()

    if not data:
        print("Belum ada data.")
        return

    total = 0

    for mahasiswa in data:
        total += mahasiswa["ipk"]

    rata_rata = total / len(data)

    print("Rata-rata IPK:", round(rata_rata, 2))


def ipk_tertinggi():
    data = baca_data()

    if not data:
        print("Belum ada data.")
        return

    tertinggi = max(data, key=lambda x: x["ipk"])

    print("\nMahasiswa dengan IPK tertinggi:")
    print("Nama :", tertinggi["nama"])
    print("NIM  :", tertinggi["nim"])
    print("IPK  :", tertinggi["ipk"])
```

```python
%%writefile student-management/main.py

from mahasiswa import (
    tambah_mahasiswa,
    lihat_mahasiswa,
    cari_mahasiswa
)

from analisis import (
    rata_rata_ipk,
    ipk_tertinggi
)


while True:

    print("\n==============================")
    print("      STUDENT MANAGEMENT")
    print("==============================")
    print("1. Tambah mahasiswa")
    print("2. Lihat mahasiswa")
    print("3. Cari mahasiswa")
    print("4. Rata-rata IPK")
    print("5. IPK tertinggi")
    print("6. Keluar")

    pilihan = input("Pilih menu: ")

    if pilihan == "1":
        tambah_mahasiswa()

    elif pilihan == "2":
        lihat_mahasiswa()

    elif pilihan == "3":
        cari_mahasiswa()

    elif pilihan == "4":
        rata_rata_ipk()

    elif pilihan == "5":
        ipk_tertinggi()

    elif pilihan == "6":
        print("Program selesai.")
        break

    else:
        print("Pilihan tidak valid.")
```
